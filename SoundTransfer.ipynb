```python
import numpy as np
import soundfile as sf
import torch
import os
import glob
from kokoro import KPipeline
from google.colab import files

pipeline = KPipeline(lang_code='a', device='cuda')

# 2. Define the path where your files are
input_folder = '/content'
output_folder = '/content/output_wavs'

if not os.path.exists(output_folder):
    os.makedirs(output_folder)

# 3. Find all LJ files in that folder
txt_files = glob.glob(f"{input_folder}/LJ*.txt")
txt_files.sort()

print(f"Found {len(txt_files)} files in {input_folder}. Starting...")

for txt_path in txt_files:
    file_name = os.path.basename(txt_path).replace('.txt', '')

    print(f"Processing: {file_name}...", end=' ')

    with open(txt_path, 'r', encoding='utf-8') as f:
        text_content = f.read().strip()

    if not text_content:
        print("Skipped (Empty)")
        continue

    # Generate Audio
    generator = pipeline(text_content, voice='af_heart', speed=1)
    all_chunks = []

    for _, (_, _, audio) in enumerate(generator):
        if torch.is_tensor(audio):
            audio = audio.cpu().numpy()
        all_chunks.append(audio)

    if all_chunks:
        final_audio = np.concatenate(all_chunks)
        sf.write(f"{output_folder}/{file_name}.wav", final_audio, 24000)
        print("Done!")

# 4. Zip and Download
print("\n--- All Finished! Zipping for download ---")
!zip -j -r results_1_50.zip {output_folder}/*.wav
files.download('results_1_50.zip')
```

```python
import os
import glob
import librosa
import soundfile as sf

INPUT_DIR  = "/mnt/c/Users/Caillou/PycharmProjects/PythonProject/dataset/audio/fake/LJSpeech-Kokoro-82M"
OUTPUT_DIR = "/mnt/c/Users/Caillou/PycharmProjects/PythonProject/dataset/audio/fake/LJSpeech-Kokoro-82M_chunked"
SAMPLE_RATE = 16000
CHUNK_DURATION = 10  # seconds

os.makedirs(OUTPUT_DIR, exist_ok=True)

files = []
for ext in ["*.wav", "*.mp3", "*.flac"]:
    files.extend(glob.glob(os.path.join(INPUT_DIR, "**", ext), recursive=True))

print(f"Found {len(files)} files to split")

total_chunks = 0
for filepath in files:
    try:
        audio, sr = librosa.load(filepath, sr=SAMPLE_RATE, mono=True)
    except Exception as e:
        print(f"[WARN] Skipping {filepath}: {e}")
        continue

    chunk_len = SAMPLE_RATE * CHUNK_DURATION
    base_name = os.path.splitext(os.path.basename(filepath))[0]

    for i, start in enumerate(range(0, len(audio) - chunk_len + 1, chunk_len)):
        chunk = audio[start:start + chunk_len]
        out_path = os.path.join(OUTPUT_DIR, f"{base_name}_chunk{i:04d}.wav")
        sf.write(out_path, chunk, SAMPLE_RATE)
        total_chunks += 1

print(f"Done! {total_chunks} chunks saved to {OUTPUT_DIR}")
```
